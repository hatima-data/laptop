```python
df=df.drop(columns=['Unnamed: 0'])
```

```python
text_columns=['Processor_variant','RAM_type', 'Graphics_name','Graphics_integreted','Graphics_brand']
df[text_columns]=df[text_columns].fillna('N/A')
```

```python
nums_cols=['Processor_gen', 'Core_per_processor', 'Total_processor', 'Execution_units', 'Threads','Graphics_GB']
df[nums_cols]=df[nums_cols].fillna(0)
```

```python
import matplotlib.pyplot as plt
import seaborn

plt.figure(figsize=(12,6))
brand_price_sum=df.groupby('Brand')['Price'].sum().sort_values(ascending=False)
brand_price_sum.plot(kind='bar', color='teal')

plt.title('Total Price of Loptops by Brand', fontsize=14, fontweight='bold')
plt.xlabel('Brand', fontsize=12)
plt.ylabel('Total Price', fontsize=12)

plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```
